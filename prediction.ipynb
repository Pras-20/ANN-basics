```python
## loading all the necessary models 

import pandas as pd 
import numpy as np 
import tensorflow  as tf 
from tensorflow.keras.models import load_model
import pickle 
```

```python
model=load_model('model.keras')
```

```python
## LOAD ALL THE NECESSARY PICKLE FILES NOW 

with open('label_encoder_gender.pkl','rb') as file:
    label_encoder_gender=pickle.load(file)
with open('OHE_country.pkl','rb') as file:
    OHE_country=pickle.load(file)
with open('scaler.pkl','rb') as file:
    scaler=pickle.load(file)
```

```python
input_data={
    'CreditScore':619,
    'Geography':'France',
    'Gender':'Female',
    'Age':42,
    'Tenure':2,
    'Balance':0,
    'NumOfProducts':1,
    'HasCrCard':1,
    'IsActiveMember':1,
    'EstimatedSalary':101348.88,
}

```

```python
geo_encoded=OHE_country.transform([[input_data['Geography']]])
```

```python
d_df=pd.DataFrame(geo_encoded,columns=OHE_country.get_feature_names_out(['Geography']))
```

```python
input_df=pd.DataFrame([input_data])
```

```python
final_df_data=pd.concat([input_df.reset_index(drop=True),d_df],axis=1)
```

```python
final_df_data.head()
```

```python
final_df_data['Gender']=label_encoder_gender.transform(final_df_data['Gender'])
```

```python
final_df_data=final_df_data.drop('Geography',axis=1)
```

```python
scaled_input=scaler.transform(final_df_data)
```

```python
predict=model.predict(scaled_input)
```
